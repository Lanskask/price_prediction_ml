```python
import datetime
```

```python
mnist = tf.keras.datasets.mnist

(X_train, y_train), (X_test, y_test) = mnist.load_data()
```

```python
import tensorflow.keras.layers as lrs
```

```python
def create_model():
    return tf.keras.models.Sequential([
      lrs.Input(shape=(28,28), name="layers_input"),
      lrs.Flatten(name="layers_flatten"),
      lrs.Dense(512, activation='relu', name="layers_dense"),
      lrs.Dropout(0.2, name="layers_dropout"),
      lrs.Dense(10, activation="softmax", name="layers_dense_2")
    ])
```

```python
model = create_model()
model.compile(optimizer="adam",
              loss="sparse_categorical_crossentropy",
              metrics=["accuracy"])

log_dir = "logs/fit" + datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
tensorboard_callback = tf.keras.callbacks.TensorBoard(log_dir=log_dir, histogram_freq=1)

model.fit(x=X_train, y=y_train,
          epochs=5, validation_data=(X_test, y_test),
          callbacks=[tensorboard_callback])
```
