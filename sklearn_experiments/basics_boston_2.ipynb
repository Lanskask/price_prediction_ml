```python
import numpy as np
from sklearn.datasets import fetch_california_housing
```

```python
housing = fetch_california_housing(as_frame=True)
```

```python
housing_df = housing.frame
```

```python
housing_df.keys()
```

```python
target_col_name = "MedHouseVal"
```

```python
X = housing_df.drop(columns=target_col_name)
```

```python
y = housing_df[target_col_name]
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
```

```python
lr.fit(X, y)
```

```python
print(f"Model coefficients (first 5): {lr.coef_[:5]}")
```

```python
pred = lr.predict(X)
```

```python
housing_df["pred"] = pred
```

```python
housing_df.keys()
```

```python
housing_df[["MedHouseVal", "pred"]].sample(5)
```

```python
housing_df["diff"] = housing_df["MedHouseVal"] - housing_df["pred"]
housing_df[["MedHouseVal", "pred","diff"]].sample(5)
```

```python
# avarage diff
average_diff = housing_df["diff"].sum() / housing_df.shape[0]
average_diff
```

```python
# Mean Abs Error
```

```python
housing_df["abs_diff"] = abs(housing_df["MedHouseVal"] - housing_df["pred"])
housing_df[["MedHouseVal", "pred","diff", "abs_diff"]].sample(5)
```

```python
av_abs_diff = housing_df["abs_diff"].sum() / housing_df.shape[0]
av_abs_diff
```

```python
from sklearn.metrics import mean_absolute_error
mean_abs_err = mean_absolute_error(housing_df[target_col_name], housing_df["pred"])
mean_abs_err
```

```python
av_abs_diff - mean_abs_err
```

```python
# mean Square Error
```

```python
housing_df["square_diff"] = (housing_df[target_col_name] - housing_df["pred"]) ** 2
housing_df[["MedHouseVal", "pred","diff", "abs_diff", "square_diff"]].sample(5)
```

```python
av_square_diff = housing_df["square_diff"].sum() / housing_df.shape[0]
av_square_diff
```

```python
from sklearn.metrics import mean_squared_error
mean_square_err = mean_squared_error(housing_df[target_col_name], housing_df["pred"])
mean_square_err
```

```python
av_square_diff - mean_square_err
```

```python
# Root out of squared mean err
```

```python
import numpy as np
```

```python
housing_df["root_square_diff"] = np.sqrt((housing_df[target_col_name] - housing_df["pred"]) ** 2)
housing_df[["MedHouseVal", "pred","diff", "abs_diff", "square_diff", "root_square_diff"]].sample(5)
```

```python
av_root_square_diff = housing_df["root_square_diff"].sum() / housing_df.shape[0]
av_root_square_diff
```

```python
from sklearn.metrics import root_mean_squared_error
root_mean_square_err = root_mean_squared_error(housing_df[target_col_name], housing_df["pred"])
root_mean_square_err
```

```python
np.sqrt(mean_square_err)
```

 R^2 = 1 - \frac { \sum_{i=1}^{n} ( y_i - \hat{y}_i )^2 }
    {\sum_{i=1}^{n} (y_i - \bar{y})^2 }

The Coefficient of Determination, $R^2$, is calculated as:

$$R^2 = 1 - \frac{\sum_{i=1}^{n} (y_i - \hat{y}_i)^2}{\sum_{i=1}^{n} (y_i - \bar{y})^2}$$

```python
1 - (housing_df[target_col_name] - housing_df["pred"]) ** 2 / (housing_df[target_col_name] - housing_df[target_col_name].mean()) ** 2
```

```python
from sklearn.metrics import r2_score
```

```python
r_squared = r2_score(housing_df[target_col_name], housing_df["pred"])
r_squared
```

```python
r_squared_by_lr = lr.score(X, y)
r_squared_by_lr
```
