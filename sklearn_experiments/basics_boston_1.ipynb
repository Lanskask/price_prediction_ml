```python
from typing import List, Union
def concat_show_all(*dfs: pd.DataFrame) -> pd.DataFrame:
  res_df = pd.concat(list(dfs), axis=1)
  with pd.option_context('display.max_columns', None):
    sample_df = res_df.sample(n=5)

  return sample_df
```
