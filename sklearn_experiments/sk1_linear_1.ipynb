```python
import numpy as np
from nibabel.cifti2 import LabelAxis
from sklearn.datasets import make_blobs
from sklearn.linear_model import LinearRegression
```

```python
# --- Данные для обучения ---
# X должен быть 2D-массивом (требование sklearn)
# Представим, что это "стаж работы" в годах
X_train = np.array([1, 2, 3, 4]).reshape(-1, 1)

# y - это "зарплата" в тыс.
y_train = np.array([30, 50, 70, 90])
```

```python
# 1. Создаем "пустую" модель
model = LinearRegression()

# 2. ОБУЧАЕМ модель (метод .fit())
# Модель "смотрит" на X_train и y_train и ищет правило
print("Обучаем модель...")
model.fit(X_train, y_train)
print("Модель обучена!")
```

```python

# --- Данные для предсказания ---
# А что, если стаж 5 или 6 лет? (Новые данные)
X_new = np.array([[5], [6]])  # Тоже 2D-массив!

# 3. ДЕЛАЕМ ПРЕДСКАЗАНИЕ (метод .predict())
predictions = model.predict(X_new)

print(f"\nНовые данные (стаж): {X_new.ravel()}")
print(f"Предсказанные зарплаты: {predictions}")
```

```python
# Building a graph
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10,10))
# the original data points
plt.scatter(X_train, y_train, edgecolor='b', s=20, label="Data")

# Plot data fit
plt.plot(X_new, predictions, color='r', linewidth=2, label="Model Fit")

plt.xlabel("Feature (X)")
plt.xlabel("Target (y)")
plt.title("Model Fix Visualization")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
```

```python
X, y_true = make_blobs(n_samples=300, centers=3, cluster_std=1.0, random_state=43)
```

```python
X_broken = np.copy(X)
X_broken[:,1] = X_broken[:,1] * 100
```

```python
km = KMeans(n_clusters=3, random_state=42, n_init=10)
```

```python
km.fit(X_broken)
labels = km.labels_
```

```python
plt.scatter(X[:,0], X[:,1], c=labels, s=50, cmap="viridis")
plt.title("Clusterization result (Bad)")
plt.show()
```
